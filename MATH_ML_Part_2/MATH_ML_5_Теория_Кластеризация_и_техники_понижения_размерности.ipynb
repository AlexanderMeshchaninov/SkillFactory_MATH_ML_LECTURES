# Оценка качества кластеризации

К сожалению, оценить качество кластеризации сложно, поскольку, в отличие от регрессии или классификации, при кластеризации у нас не всегда есть эталонные ответы, с которыми можно сравнить результаты предсказаний!

Однако, выделяют две больших группы методов, с помощью которых можно оценить результат кластеризации:

1. __Внутренние меры__ — это метрики, которые оценивают качество кластеризации только по той информации, которая есть в неразмеченных данных.

2. __Внешние меры__ — это метрики, которые используют, если в данных есть информация о некотором заранее известном разделении на кластеры, с которым можно сравнить полученный результат.

# Внутренние меры

В библиотеке sklearn реализованы три наиболее популярные метрики:

- коэффициент силуэта (Silhouette Coefficient);
- индекс Калински — Харабаса (Calinski-Harabasz Index);
- индекс Дэвиса — Болдина (Davies-Bouldin Index).

# Коэффициент силуэта

При решении практических задач вы будете использовать уже готовые алгоритмы на Python. Для вычисления коэффициента силуэта в библиотеке sklearn используется silhouette_score:

```python
    # определяем алгоритм кластеризации
    km = KMeans(n_clusters=3, random_state=42)
    # обучаем его на наших данных
    km.fit_predict(X)
    # вычисляем значение коэффициента силуэта
    score = silhouette_score(X, km.labels_, metric='euclidean')
```

Значение коэффициента силуэта всегда находится в диапазоне [−1,1].

- Значение близко к −1: объекты в кластерах разрознены, и в целом кластерную структуру не удалось выделить.
- Значение близко к 0: кластеры пересекаются друг с другом.
- Значение близко к 1: чёткая кластерная структура с «плотными», отделёнными друг от друга кластерами.

Иными словами, чем выше значение коэффициента силуэта (ближе к 1), то объекты кластера лучше вписываются в сам кластер, так же насколько он далек от других кластеров.

# Индекс Калински — Харабаса

Следующий коэффициент, который мы рассмотрим, — это индекс Калински — Харабаса. Он показывает отношение между разбросом значений между кластерами и разбросом значений внутри кластеров.

Иными словами, индекс оценивает качество кластеризации на основе разделённости между кластерами и компактности внутри кластеров.

Чем выше значение индекса, тем лучше кластеризация:

- кластеры более плотные внутри,
- хорошо разделены между собой.

В библиотеке sklearn данный алгоритм реализуется с помощью метода calinski_harabasz_score():

```python
    # определяем алгоритм кластеризации
    km = KMeans(n_clusters=3, random_state=42)
    # обучаем его на наших данных
    km.fit_predict(X)
    # вычисляем значение коэффициента Калински — Харабаса
    score = calinski_harabasz_score(X, km.labels_)
```

# Индекс Дэвиса — Болдина

Это метрика качества кластеризации, которая оценивает, насколько кластеры компактны и хорошо разделены.

Чем ниже значение индекса, тем лучше кластеризация:

- кластеры более плотные внутри,
- и дальше друг от друга.

В библиотеке sklearn индекс Дэвиса — Болдина реализуется с помощью метода davies_bouldin_score():

``` python
    # определяем алгоритм кластеризации
    km = KMeans(n_clusters=3, random_state=42)
    # обучаем его на наших данных
    km.fit_predict(X)
    # вычисляем значение коэффициента Дэвиса — Болдина
    score = davies_bouldin_score(X, km.labels_)
```

<img src='Images/math_ml_40.png'>

# Внешние меры

Основные внешние метрики:

- Индекс Рэнда и Скорректированный индекс Рэнда (Adjusted Rand Index) — учитывает случайные совпадения, значение от -1 до 1 (чем ближе к 1 — тем лучше).
- Нормализованная взаимная информация (Normalized Mutual Information) — измеряет общую информацию между кластерами и истинными метками (от 0 до 1).
- Однородность, Полнота (Homogeneity, Completeness, V-measure) — оценивают, насколько кластеры однородны и полны.

# Индекс Рэнда

Индекс Рэнда — это мера сходства между двумя кластеризациями.

Для вычисления индекса Рэнда с помощью библиотеки sklearn можно использовать метод rand_score().

Для вычисления скорректированного индекса Рэнда в sklearn представлен через метод adjusted_rand_score().

Интерпретация скорректированного индекса Рэнда:

- Отрицательные значения индекса говорят о различных, независимых разбиениях.
- Значения, близкие к нулю, являются индикаторами случайного разбиения.
- Положительные значения показывают схожесть разбиений (значение, равное 1, получается при полном совпадении кластеризаций).

# Нормализованная взаимная информация

В sklearn для вычисления NMI существует метод normalized_mutual_info_score().

Интерпретация (принимает значения от 0 до 1):

- 1 — идеальное совпадение кластеров с эталоном.
- 0 — нет взаимной информации между кластеризацией и истинными метками

# Однородность

Она (гомогенность) показывает, насколько элементы в кластере похожи между собой.

Однородность также можно вычислить, воспользовавшись методом homogeneity_score() из библиотеки sklearn.

Интерпретация:

Важно, чтобы показатель гомогенности был как можно ближе к единице: это говорит о том, что в кластере находятся элементы одного и того же класса. Если он близок к нулю, это означает, что в кластере находятся элементы из разных классов.

# Полнота

Результат кластеризации удовлетворяет требованиям полноты, если все элементы данных, принадлежащие к одному классу, оказались в одном кластере.

Для вычисления полноты с помощью библиотеки sklearn используется метод completeness_score().

Интерпретация:

По аналогии с однородностью, важно, чтобы значение метрики было близко к единице.

<img src='Images/math_ml_41.png'>

# V-мера

Мы уже выяснили, что лучше рассматривать несколько мер сразу, так как каждая мера в отдельности может показывать высокое качество даже для плохого разбиения. В ответ на такую необходимость появляется метрика, которая является комбинацией метрик полноты и однородности и вычисляется как их гармоническое среднее, — это V-мера.

Интерпретация:

Значение V-меры варьируется от 0 до 1. Метрика будет равна  только в том случае, если кластеры будут однородными и полными одновременно.

Для вычисления V-меры с помощью библиотеки sklearn используется метод v_measure_score().

<img src='Images/math_ml_42.png'>

ВАЖНО:

Стоит отметить, что у всех перечисленных метрик есть большой недостаток: для их использования необходимо знать истинные принадлежности к кластерам. На практике это практически всегда недоступно (а если разметка всё же есть, то проще перейти к задаче классификации) или требует финансовых и трудовых затрат, к примеру, если мы делаем разметку вручную.

# PCA (Principal Component Analysis, метод главных компонент)

Размерность данных отражает количество содержащихся в них признаков. Если признаков очень много, можно столкнуться с проклятием размерности.

__Проклятие размерности__ — это проблема, связанная с экспоненциальным возрастанием объёма данных из-за увеличения размерности пространства.

Основные подходы к снижению размерности:

1. __Удаление признаков__ может быть подходящим вариантом, когда, например, в них есть некорректные данные или так много пропусков, что эту информацию нельзя использовать для построения модели. Иногда удаляют малоинформативные признаки, которые принимают одни и те же значения или у которых очень слабая корреляция с целевой переменной. Есть и более сложные методы фильтрации — сейчас мы не будем на них останавливаться, однако обратим ваше внимание, что при удалении признаков мы практически всегда теряем какую-то информацию о данных.

2. __Создание новых признаков__ не даёт потерять важную информацию. Допустим, в нашем наборе данных есть 100 признаков. При извлечении признаков мы отказываемся от старых и создаём 100 новых независимых (таких, между которыми коэффициент корреляции равен нулю). Каждый новый признак представляет собой комбинацию старых. Независимые новые признаки определяются специальным алгоритмом (например, PCA) и упорядочиваются по тому, насколько хорошо они сохраняют информацию о данных. После этого можно отбросить наименее важные новые признаки в зависимости от того, какую объяснённую дисперсию необходимо сохранить: допустим, первые 80 признаков из извлечённых 100 объясняют 95% дисперсии и нас устраивает такой процент. Значит, мы отбрасываем последние 20 новых признаков.

В результате множества исследований было выявлено, что такое отбрасывание части новых признаков совершенно не мешает получать модели машинного обучения, которые будут показывать высокую эффективность. Например, для распознавания лиц достаточно использовать лишь треть извлечённых из фотографий признаков, а для обнаружения лактозы в безлактозном молоке NIR-спектроскопией будет достаточно извлечь с помощью PCA всего лишь два признака из 601.

# Алгоритм реализации PCA:

1. Стандартизировать данные.

2. Рассчитать ковариационную матрицу для объектов.

3. Рассчитать собственные значения и собственные векторы для ковариационной матрицы.

4. Отсортировать собственные значения и соответствующие им собственные векторы.

5. Выбрать k наибольших собственных значений и сформировать матрицу соответствующих собственных векторов.

6. Преобразовать исходные данные, умножив матрицу данных на матрицу отобранных собственных векторов.

<img src='Images/math_ml_43.png'>

# В python

Данный алгоритм можно реализовать с помощью библиотеки sklearn

``` python
    # импортируем нужный алгоритм
    from sklearn.decomposition import PCA
    # определяем метод главных компонент с двумя компонентами
    pca = PCA(n_components=2)
    # обучаем алгоритм на наших данных
    principalComponents = pca.fit_transform(df_std)
```

# SVD

SVD — это способ разложить любую матрицу на три специальные матрицы, которые дают много полезной информации о исходной матрице. Это очень мощный инструмент в линейной алгебре, который широко используется в машинном обучении, анализе данных, обработке изображений и других областях.

### Как это работает?
Пусть у нас есть матрица A размером m x n (то есть с m строками и n столбцами). С помощью SVD мы можем представить A так: $A = UΣV^T$

Что это значит?

* U — это матрица размера m x m, которая состоит из так называемых левых сингулярных векторов. Она ортогональна, то есть её столбцы взаимно перпендикулярны и имеют длину 1.

* Σ — диагональная матрица размера m x n, на диагонали которой расположены сингулярные значения. Эти числа всегда неотрицательные и идут по убыванию — от большого к маленькому. Остальные элементы — нули.

* $V^T$  — транспонированная матрица V размером n x n, которая содержит правые сингулярные вектор (аналогично U, её столбцы ортогональны).

- __Сингулярные значения__ показывают "важность" или "вес" каждого сингулярного вектора — чем больше значение, тем сильнее влияние соответствующего компонента на матрицу A.

- Благодаря такому разложению можно выделить основные направления, вдоль которых данные наиболее значимы, а менее важные можно отбросить, что помогает уменьшить размерность данных.

### Где применяют SVD?

1. Снижение размерности (PCA):

Если у вас много признаков, SVD помогает найти самые важные из них и уменьшить количество переменных без сильной потери информации.

2. Рекомендательные системы:

В системах, например, для рекомендаций фильмов или товаров, SVD помогает выявить скрытые связи между пользователями и товарами.

3. Обработка изображений:

Можно сжать изображение, оставив только главные сингулярные значения, сохранив при этом его основную структуру.

4. Решение систем уравнений и устранение шума:

В задачах с переопределёнными или плохо обусловленными системами SVD помогает найти наилучшее приближение решения.

### Пример:

Представьте, что у вас есть таблица с оценками пользователей для фильмов (например, 5 пользователей и 4 фильма). SVD разложит эту таблицу так, что вы сможете понять:

- Какие основные "типы" пользователей есть (например, любители боевиков, комедий и т.д.)

- Какие основные "типы" фильмов существуют

-0 Как сильно каждый тип пользователя связан с каждым типом фильма

Это помогает делать более точные рекомендации.

### Кратко про матрицы U,Σ,V:

- U — описывает, как строки исходной матрицы связаны с основными компонентами.
- Σ — показывает важность каждой компоненты.
- V — описывает, как столбцы исходной матрицы связаны с основными компонентами.

Для того чтобы реализовать сингулярное разложение с помощью библиотеки sklearn, необходимо использовать алгоритм TruncatedSVD(), в который передаётся n_components в качестве параметра, определяющего количество итоговых компонент:

``` python
    # создаём объект класса TruncatedSVD
    # n_components — размерность нового пространства, n_iter — количество итераций
    svd = TruncatedSVD(n_components=5, n_iter=7, random_state=42)
    # обучаем модель на данных X
    svd.fit(X)
    # применяем уменьшение размерности к матрице X
    transformed = svd.transform(X)
```

# t-SNE

__t-SNE (t-distributed Stochastic Neighbor Embedding)__ — это метод снижения размерности, который используется для визуализации сложных данных в 2D или 3D. Он особенно хорошо показывает группы (кластеры) внутри данных. Его преимущество заключается в том, что он может реализовывать уменьшение размерности и разделение для данных, которые являются __линейно неразделимыми__ (нелинейно уменьшает размерности).

### Зачем нужен t-SNE?

Когда у тебя есть данные с большим числом признаков (например, 50, 100, 1000+), их невозможно визуализировать напрямую. t-SNE помогает "сжать" такие данные в 2D или 3D, сохранив структуру: какие точки рядом, какие — далеко.

### Где используется?

- В машинном обучении — для анализа признаков и оценки качества кластеризации

- В обработке изображений — чтобы визуализировать похожие изображения

- В биоинформатике — для анализа данных генов

- В обработке текста и речи — для отображения эмбеддингов слов или предложений

Для реализации t-SNE в sklearn понадобится алгоритм TSNE():

``` python
    # импортируем класс TSNE из модуля manifold библиотеки sklearn
    from sklearn.manifold import TSNE
    # создаём объект класса TSNE
    # n_components — размерность нового пространства
    tsne = TSNE(n_components=2, perplexity=30, n_iter=500, random_state=42)
    # обучаем модель на данных X и применяем к матрице X уменьшение размерности
    tsne.fit_transform(X)
```

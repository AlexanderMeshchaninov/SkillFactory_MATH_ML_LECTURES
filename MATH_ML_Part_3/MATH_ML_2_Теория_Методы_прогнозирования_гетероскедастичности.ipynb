# Модели прогнозирования гетероскедастичности

Представьте, что мы строим прогноз, а потом смотрим, насколько ошиблась модель в разные моменты времени.
Если в начале ошибки маленькие, а потом становятся всё больше и больше, или наоборот — это и есть __гетероскедастичность__.

<img src='Images/math_ml_17.png'>

Такой характер дисперсии можно часто встретить __во временных рядах из финансовой сферы__, поскольку они более подвержены гетероскедастичности.

__Отсутствие__ гетероскедастичности называется __гомоскедастичностью__.

<img src='Images/math_ml_18.png'>

Поскольку модели, которые мы тут использовали не умеют улавливать дисперсию во времени их предсказания могут быть неверными!

Для данных с непостоянной дисперсией был разработан ещё один класс моделей — семейство моделей авторегрессионной условной гетероскедастичности, или __ARCH (Autoregressive Conditional Heteroscedastic Model) и GARCH (Generalized Autoregressive Conditional Heteroscedastic Model)__.

__ARCH и GARCH__ — наиболее популярные модели, используемые при прогнозе волатильности. Существуют и их модернизации, такие как A-GARCH, E-GARCH и многие другие. В будущем, если вам придётся тесно работать с финансовыми временными рядами, вы сможете самостоятельно познакомиться с этими моделями.

# ARCH

Модель, которая говорит:

«Если в прошлом были большие скачки (высокая изменчивость), то в будущем тоже может быть нестабильно».

Если ошибка была большой, то модель ожидает высокую волатильность.

ARCH полезна, когда у нас в данных есть "вспышки" нестабильности — то тихо, то шторм.

# GARCH

Улучшенная версия ARCH

Добавляет:

Не только прошлые ошибки, но и прошлую волатильность (т.е. "если вчера была буря, скорее всего и сегодня тоже будет неспокойно")

Просто:

GARCH = ARCH + память о недавней волатильности

# Когда применять ARCH и GARCH?

* Когда ряд похож на белый шум, но при этом в нём присутствует гетероскедастичность.
Чтобы определить, является ли ряд гетероскедастичным (с меняющейся дисперсией), можно отобразить его квадраты на графике и понаблюдать за поведением дисперсии.

* Когда после применения AR-модели остатки (ошибки модели) тоже являются гетероскедастичными. В этом случае вы также можете прогнозировать дисперсию ошибок и использовать её в итоговом предсказании. Для этого необходимо суммировать результаты AR-модели с результатами ARCH.

# Волатильность

__Волатильность__ — это то, насколько "нервно" или "рвано" ведёт себя временной ряд.

* Если значения меняются плавно, волатильность низкая.

* Если значения скачут резко вверх и вниз, волатильность высокая.

Зачем это нужно?

Чтобы понять, насколько стабильны изменения во времени, и можно ли на них положиться при прогнозе!

Особенно важно в таких задачах:

* Финансы (цены акций, валюты)

* Инфляция, курсы

* Продажи (если бывают резкие всплески спроса)

* Цены на нефть, металлы, энергию

### Пример:

Допустим, курс доллара:

* В январе — 90₽

* В феврале — 93₽

* В марте — 120₽

Вот тут __волатильность высокая__ — нужно быть __осторожным в прогнозах__.

### Какие ряды проверять?

1. Ряды с экономическими значениями	- часто нестабильны по природе
2. Финансовые ряды (цены, акции) - почти всегда имеют меняющуюся волатильность
3. Продажи с всплесками спроса - могут иметь временные "бури".
